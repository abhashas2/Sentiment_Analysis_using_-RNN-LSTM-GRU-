```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=['I love my dog', 'I love my cat', 'You love my dog!', 'Do you think my dog is amazing?']
```

```python
sent
```

```python
## Define the vocabulary size
vocab_size=10000
```

```python
## one hot representation for every word in the sentence
onehot_repr=[one_hot(words,vocab_size)for words in sent]
onehot_repr
```

```python
## word embediding representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(onehot_repr,padding='pre',maxlen=sent_length)
embedded_docs
```

```python
## feature representation using embedding
dim=10
```

```python
model=Sequential()
model.add(Embedding(vocab_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

```python
model.predict(np.array([embedded_docs[0]]))
```

## END TO END SIMPLE RNN PROJECT


```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN ,Dense, Dropout
```

```python
## Load the imdb dataset
max_features=10000 ## vocabulary_size
(X_train, y_train), (X_test, y_test) = imdb.load_data(
    num_words=max_features
)

# print the shape of the data
print(f'training data shape: {X_train.shape}, Training labels shape: {y_train.shape}')
print(f'testing data shape: {X_test.shape}, Testing labels shape: {y_test.shape}')
```

```python
## Inspect a sample review and its label
#X_train[0]
```

```python
y_train[0]
```

```python

sample_review=X_train[0]
sample_label=y_train[0]
print(f'Sample review (as integer):{sample_review}')
print(f'Sample label (as integer):{sample_label}')
```

```python
## Mapping of words index back to word(for understanding)

word_index=imdb.get_word_index()
```

```python
max_len=500
X_train = sequence.pad_sequences(X_train, maxlen=max_len)
X_test = sequence.pad_sequences(X_test, maxlen=max_len)
X_train
```

```python
model = Sequential()

model.add(
    Embedding(
        input_dim=max_features,
        output_dim=128,
        input_length=max_len
    )
)

model.add(
    SimpleRNN(
        128,
        activation='tanh'
    )
)

model.add(
    Dropout(0.3)
)

model.add(
    Dense(
        1,
        activation='sigmoid'
    )
)

model.build(input_shape=(None, max_len))
```

```python
model.summary()
```

```python
## Create an isinstance of earlystoppingg
from tensorflow.keras.callbacks import EarlyStopping
EarlyStopping=EarlyStopping(monitor='val_loss', patience=5, restore_best_weights= True)
EarlyStopping
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001,
        clipnorm=1.0
    ),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

History = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
   callbacks=[EarlyStopping]
)
```

```python
model.save('SIMPLE_RNN_imdb.h5')
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test
)

print("Test Accuracy:", test_accuracy)
```
